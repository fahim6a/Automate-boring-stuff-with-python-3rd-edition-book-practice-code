4 ways to do string formatting. the format specifier %s for string, %d for decimal, %f for floating and %x supported like c 

there is a big problem in old ways format specifier. easy to get messed up on small changes, what if you swap the (key, vlaues) to (values, key). get error. 

```python
old_ways = '%-10s = %.2f' % (team, time)
```

in new_ways just make sure that, format string placeholder name and string placeholder name exactly same like %(key):key


```python
new_ways = '%(team)-10s = %(time).2f' %{'team':team, 'time': time}
```

```python
print(show_message)
```

```python
print(new_ways)
```
