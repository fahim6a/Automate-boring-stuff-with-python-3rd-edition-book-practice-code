write a first function that bytes or string as instance and always return a str 

```python
def to_str(bytes_or_str):
    if isinstance(bytes_or_str, bytes): #
        value = bytes_or_str.decode('utf-8')
    return value
```

```python
print(repr(to_str(b'england'))) #repr() is a representation of a string, b represent bytes here. 
```

The second function write you down take bytes or str and always return bytes

```python
def to_bytes(bytes_or_str):
    if  isinstance(bytes_or_str, str):
        value = bytes_or_str.encode('utf-8')
        return value
```
