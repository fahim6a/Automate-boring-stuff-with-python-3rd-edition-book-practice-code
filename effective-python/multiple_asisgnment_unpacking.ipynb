swap the values between two position in a list a part of the accending order 

```python
def bubble_sort(a):
    for _ in range(len(a)):
        for i in range(1, len(a)):
            if a[i] < a[i-1]:
                temp = a[i]
                a[i]= a[i-1]
                a[i-1]=temp 

names=['fahim','sajal', 'harry','amir']
bubble_sort(names)
print(names)
```
